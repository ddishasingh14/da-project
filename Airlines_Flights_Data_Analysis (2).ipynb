```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
data = pd.read_csv('/content/drive/MyDrive/Data_Analytics/airlines_flights_data.csv')
display(data.head())
```

```python
data.head()
```

```python
data.info()
```

```python
data.drop(columns='index', inplace=True)
data.head()
```

```python
df_airlines = data['airline'].value_counts().reset_index()
df_airlines
```

```python
norm = plt.Normalize(df_airlines["count"].min(), df_airlines["count"].max())
colors = plt.cm.Blues(norm(df_airlines["count"]))

plt.figure(figsize=(10,7))
sns.barplot(data=df_airlines, x='airline', y='count', palette=colors)
plt.title('Airlines with Frequencies')
plt.xlabel('Airlines')
plt.ylabel('Frequencies')
plt.tight_layout()
plt.show()
```

```python
order = ['Early_Morning', 'Morning', 'Afternoon', 'Evening','Night', 'Late_Night']

fig, axes = plt.subplots(1, 2, figsize=(15,6))

sns.countplot(data=data,
              x='departure_time',
              palette='Set2',
              order=order,
              ax=axes[0])
axes[0].set_title('Departure Time')
axes[0].set_xlabel('Departure Time')
axes[0].set_ylabel('Frequencies')

sns.countplot(data=data,
              x='arrival_time',
              palette='Set2',
              order=order,
              ax=axes[1])
axes[1].set_title('Arrival Time')
axes[1].set_xlabel('Arrival Time')
axes[1].set_ylabel('Frequencies')

plt.tight_layout()
plt.show()
```

```python
data['source_city'].value_counts()
```

```python
order = ['Delhi', 'Mumbai', 'Bangalore', 'Kolkata', 'Hyderabad', 'Chennai']

fig, axes = plt.subplots(1, 2, figsize=(15,6))

ax0 = sns.countplot(data=data,
              x='source_city',
              palette='Set3',
              order=order,
              ax=axes[0])
for p in ax0.patches:
    height = p.get_height()
    ax0.annotate(f"{height}",
                 (p.get_x() + p.get_width() / 2., height),
                 ha='center', va='bottom', fontsize=10)

ax1 = sns.countplot(data=data,
              x='destination_city',
              palette='Set3',
              order=order,
              ax=axes[1])
for p in ax1.patches:
    height = p.get_height()
    ax1.annotate(f"{height}",
                 (p.get_x() + p.get_width() / 2., height),
                 ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
data.groupby(['airline', 'class'])['price'].mean().reset_index()
```

```python
plt.figure(figsize=(10, 7))
sns.boxplot(data=data, x='airline', y='price', hue='class')
plt.title('Prices Variety by Airlines')
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 7))

order = ['Early_Morning', 'Morning', 'Afternoon', 'Evening', 'Night', 'Late_Night']
sns.boxplot(data=data,
            x='departure_time',
            y='price',
            order=order,
            palette='Set3',
            ax=axes[0])
axes[0].set_title('Price by Departure Time')

sns.boxplot(data=data,
            x='arrival_time',
            y='price',
            order=order,
            palette='Set3',
            ax=axes[1])
axes[1].set_title('Price by Arrival Time')

plt.tight_layout()
plt.show()
```

```python
order = ['Early_Morning', 'Morning', 'Afternoon', 'Evening', 'Night', 'Late_Night']
pivot = data.pivot_table(values='price',
                       index='departure_time',
                       columns='arrival_time',
                       aggfunc='mean').reindex(index=order, columns=order)

plt.figure(figsize=(12,7))

sns.heatmap(pivot, annot=True, fmt='.0f', cmap='Blues')
plt.title('Average Ticket Price by Departure and Arrival Time')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 7))

order = ['Bangalore', 'Chennai', 'Delhi', 'Hyderabad', 'Kolkata', 'Mumbai']
sns.boxplot(data=data,
            x='source_city',
            y='price',
            order=order,
            palette='Set2',
            ax=axes[0])
axes[0].set_title('Price Based on Source City')

sns.boxplot(data=data,
            x='destination_city',
            y='price',
            order=order,
            palette='Set2',
            ax=axes[1])
axes[1].set_title('Price Based on Destination City')

plt.tight_layout()
plt.show()
```

```python
order = ['Bangalore', 'Chennai', 'Delhi', 'Hyderabad', 'Kolkata', 'Mumbai']
pivot = data.pivot_table(values='price',
                       index='source_city',
                       columns='destination_city',
                       aggfunc='mean').reindex(index=order, columns=order)

plt.figure(figsize=(12, 7))

sns.heatmap(pivot, annot=True, fmt='.0f', cmap='Blues')
plt.title('Average Ticket Price by Source City and Destination Sity')
plt.show()
```
